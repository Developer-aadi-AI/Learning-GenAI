```python
from fastmcp import FastMCP
import threading
import time

mcp = FastMCP("Calculator MCP")

@mcp.tool()
def add(a: int, b: int) -> int:
    return a + b

@mcp.tool()
def add(a: int, b: int) -> int:
    return a*b

def run_server():
    mcp.run(transport = "http") # Important

threading.Thread(target = run_server, daemon = True).start()

time.sleep(2)
print("MCP HTTP server running!")
```

```python
from fastmcp import Client

async def main():
    async with Client("http://127.0.0.1:8000/mcp") as client:
        result = await client.call_tool("add", {"a" : 100, "b" : 50})
        print(result)

await main()
```
